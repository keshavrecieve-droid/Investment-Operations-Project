```python
#Load everything
import pandas as pd

positions = pd.read_excel(r"C:\Keshav S\Investements Operations Platform\Closing_Positions.xlsx")
prices = pd.read_excel(r"C:\Keshav S\Investements Operations Platform\Prices.xlsx")
cash = pd.read_excel(r"C:\Keshav S\Investements Operations Platform\Cash.xlsx")
```

```python
#Recreate Portfolio Value
portfolio = positions.merge(prices, on="Security", how="left")

portfolio["Market_Value"] = (
    portfolio["Closing_Position"] * portfolio["Market_Price"]
)

total_portfolio_value = portfolio["Market_Value"].sum()
print("Portfolio Value:", total_portfolio_value)
```

```python
#Cash Value
cash_balance = cash["Opening_Cash"].iloc[0]
print("Cash Balance:", cash_balance)
```

```python
#Total AUM (very important KPI)
total_aum = total_portfolio_value + cash_balance
print("Total AUM:", total_aum)
```

```python
#Exposure Table
portfolio["Exposure_%"] = (
    portfolio["Market_Value"] / total_portfolio_value * 100
)
```

```python
#Top Holdings
top_holdings = portfolio.sort_values(
    "Market_Value",
    ascending=False
).head(5)
```

```python
#Create Final KPI Report
with pd.ExcelWriter(
    r"C:\Keshav S\Investements Operations Platform\Final_KPI_Report.xlsx"
) as writer:
    
    portfolio.to_excel(writer, sheet_name="Portfolio", index=False)
    top_holdings.to_excel(writer, sheet_name="Top_Holdings", index=False)
    
    summary = pd.DataFrame({
        "Metric": [
            "Portfolio Value",
            "Cash Balance",
            "Total AUM"
        ],
        "Value": [
            total_portfolio_value,
            cash_balance,
            total_aum
        ]
    })
    
    summary.to_excel(writer, sheet_name="Summary", index=False)
```
