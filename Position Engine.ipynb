```python
#Read Files
import pandas as pd

positions = pd.read_excel(r"C:\Keshav S\Investements Operations Platform\Opening_Positions.xlsx")
trades = pd.read_excel(r"C:\Keshav S\Investements Operations Platform\Trades.xlsx")

print(positions.head())
print(trades.head())
```

```python
#Convert Sells into Negative Quantities
trades['Trade_Impact'] = trades.apply(
    lambda x: x['Quantity']
    if x['Buy_Sell'] == 'Buy'
    else -x['Quantity'],
    axis=1
)
```

```python
#Calculate Net Impact per Security
net_trades = (
    trades.groupby('Security')['Trade_Impact']
    .sum()
    .reset_index()
)
```

```python
#Merge with Opening Positions
closing_positions = positions.merge(
    net_trades,
    on='Security',
    how='left'
)
```

```python
#Calculate Closing Position
closing_positions['Trade_Impact'] = (
    closing_positions['Trade_Impact']
    .fillna(0)
)

closing_positions['Closing_Position'] = (
    closing_positions['Opening_Position']
    +
    closing_positions['Trade_Impact']
)
```

```python
#Export to Excel
closing_positions.to_excel(
    "Closing_Positions.xlsx",
    index=False
)

print("Closing Positions File Created")
```
