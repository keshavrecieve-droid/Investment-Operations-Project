```python
#Cash Movement 
#load Trades
import pandas as pd

trades = pd.read_excel(r"C:\Keshav S\Investements Operations Platform\Trades.xlsx")
cash = pd.read_excel(r"C:\Keshav S\Investements Operations Platform\Cash.xlsx")
```

```python
#Create Cash Impact
trades["Cash_Impact"] = trades.apply(
    lambda x: -x["Quantity"] * x["Trade_Price"]
    if x["Buy_Sell"] == "Buy"
    else x["Quantity"] * x["Trade_Price"],
    axis=1
)
```

```python
#Aggregate Cash Flow
cash_flow = trades.groupby("Security")["Cash_Impact"].sum().reset_index()
cash_flow.columns = ["Security", "Net_Cash_Impact"]
```

```python
#Total Cash Movement
total_cash_impact = cash_flow["Net_Cash_Impact"].sum()
print("Net Cash from Trades:", total_cash_impact)
```

```python
#Closing Cash Position
opening_cash = cash["Opening_Cash"].iloc[0]

closing_cash = opening_cash + total_cash_impact

print("Closing Cash:", closing_cash)
```

```python
#export
cash_flow.to_excel(
    r"C:\Keshav S\Investements Operations Platform\Cash_Report.xlsx",
    index=False
)
```
